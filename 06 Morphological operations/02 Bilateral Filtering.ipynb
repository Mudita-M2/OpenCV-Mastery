Unlike a normal blur, it calculates its blending weights using two factors:

Spatial Closeness: Nearby pixels have a higher influence on the output

Intensity Similarity: Pixels with a similar brightness contribute more to the blend

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
````

````python
img = cv2.imread(r'C:\Users\Hp\Pictures\OpenCVSamples\cat.webp')
img = cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
````

cv2.bilateralFilter(img, neighborhood_diameter, color_sigma, space_sigma)

bilateral = cv2.bilateralFilter(img, 15, 75, 75)

``` TEXT

| Parameter | Simple meaning                                                                     |
| --------- | ---------------------------------------------------------------------------------- |
| `img`     | Input image                                                                        |
| `15`      | **Neighborhood diameter** — how large an area around each pixel is considered      |
| `75`      | **Color sigma** (`sigmaColor`) — how much difference in color/intensity is allowed |
| `75`      | **Space sigma** (`sigmaSpace`) — how much spatial distance matters                 |


````python
bi = cv2.bilateralFilter(img,15,75,75)
plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('original image')

plt.subplot(1,2,2)
plt.imshow(bi)
plt.title("Filtered image")

plt.show()
````
