Image → Grayscale → Otsu Threshold → Invert → Morphological Gradient → Display

Morphological gradient is basically used to find the boundaries/edges of objects.

It compares a dilated version and an eroded version of the image.
```TEXT

Conceptually:

Dilation
   ↓
makes white regions bigger


Erosion
   ↓
makes white regions smaller


Difference between them
   ↓
BOUNDARY / EDGE

So instead of getting the whole object, you mainly get its outline.

````python
import cv2, numpy as np, matplotlib.pyplot as plt
img = cv2.imread(r"C:\Users\Hp\Pictures\OpenCVSamples\ABCD.jpg", 0)  
bin = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1]  

k = np.ones((3, 3), np.uint8)  
inv = cv2.bitwise_not(bin)  
out = cv2.morphologyEx(inv, cv2.MORPH_GRADIENT, k) 
plt.imshow(out, cmap='gray'), plt.axis('off'), plt.show()
````
