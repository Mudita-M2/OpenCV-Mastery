```python
import cv2
import matplotlib.pyplot as plt 
import numpy as np 
```

Phase 1: Setup and The Golden Rule
Before you can perform any bitwise operations, you must prepare your inputs according to the strict theoretical rules of OpenCV.
Define Inputs: You need two binary images (black background, white shapes)

For example, Input Image 1 could be a white rectangle, and Input Image 2 could be a white circle.
Enforce the Rule: Ensure both images have the exact same dimensions
.
Check Array Type: Ensure both inputs are loaded or created as single-channel, 8-bit (or floating-point) array

```python
img1 = cv2.imread(r'C:\Users\Hp\Pictures\OpenCVSamples\Binary_img_1.png')
img2 = cv2.imread(r'C:\Users\Hp\Pictures\OpenCVSamples\Binary_img_2.png')
```

```python
plt.figure(figsize=(12 , 6))

plt.subplot(1, 2, 1)
plt.imshow(img1)
plt.title('Image 1')    

plt.subplot(1, 2, 2)
plt.imshow(img2)        
plt.title('Image 2')

plt.show()
```

Phase 2: Extracting Overlaps (AND)
First, you will find out exactly where the two shapes intersect. 4.  The Operation: Use cv2.bitwise_and(source1, source2)
. Pass your two binary images as the sources. 5.  What to observe: The output image should be mostly black, with a white shape representing only the specific area where the white pixels of Image 1 and Image 2 overlapped
.

```python
intersection = cv2.bitwise_and(img1, img2)
plt.imshow(intersection)
```

Phase 3: Image Creation & Combination (OR)
Next, you will use bitwise logic to combine the two separate shapes into a single cohesive image mask. 6.  The Operation: Use cv2.bitwise_or(source1, source2)
. 7.  What to observe: The output will display the full white shapes from both Image 1 and Image 2 seamlessly merged together on the black background
.

```python
union = cv2.bitwise_or(img1, img2)
plt.imshow(union)
```

Phase 4: Highlighting Differences (XOR)
Now, you will extract only the parts of the shapes that do not overlap. 8.  The Operation: Use cv2.bitwise_xor(source1, source2)
. 9.  What to observe: The output will look like the OR combined image you just made, but the intersecting area in the middle will be completely punched out (turned black), leaving only the non-overlapping regions

```python
xor = cv2.bitwise_xor(img1, img2)
plt.imshow(xor)
```

Phase 5: Inversion (NOT)
Finally, you will take the combined image you created in Phase 3 and completely invert its properties. 10. The Operation: Use cv2.bitwise_not(source)
. 11. Parameter Check: Remember the common beginner mistake! Do not pass both images into this function. Pass only the single output array you generated during the bitwise_or step
. 12. What to observe: The black background will become completely white, and your combined shapes will turn completely black
.

```python
opp = cv2.bitwise_not(img1)
plt.imshow(opp)
```
