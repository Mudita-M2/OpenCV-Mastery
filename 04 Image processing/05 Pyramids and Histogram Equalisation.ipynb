```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread(r'C:\Users\Hp\Pictures\OpenCVSamples\pyramid.webp')
```

Pyramid Down with cv2.pyrDown() - downscaling - It decreases the image's size (both width and height) by a factor of 2

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
downgraded_img = cv2.pyrDown(img)

plt.imshow(img)
plt.title('Original Image')
plt.show()

plt.imshow(downgraded_img)
plt.title('Downgraded Image')
plt.show()
```

Pyramid Up with cv2.pyrUp()

This function performs the opposite operation of cv2.pyrDown(). It increases the image's size (both width and height) by a factor of 2

```python

upgraded_img = cv2.pyrUp(img)

plt.imshow(img)
plt.title('Original Image')
plt.show()

plt.imshow(upgraded_img)
plt.title('Upgraded Image')
plt.show()
```

Histogram Equalisation - Contrast Enhancement

```python
img_hist = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

equ = cv2.equalizeHist(img_hist)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_hist, cmap='gray')
plt.title('Original Image Histogram')
plt.subplot(1, 2, 2)
plt.imshow(equ, cmap='gray')
plt.title('Equalized Image Histogram')

plt.show()

```
