```python
img = cv2.imread(r'C:\Users\Hp\Pictures\OpenCVSamples\blur1.webp')
```

You should process the image in BGR first, and then convert it to RGB only when you are ready to display it.
Processing in BGR (OpenCV's Domain) When you use cv2.imread(), OpenCV loads the image into memory in the BGR format. Because OpenCV's entire ecosystem is built around this default, all of its processing functions—such as cv2.GaussianBlur(), cv2.medianBlur(), and cv2.bilateralFilter()—are designed to natively accept and process BGR array

Your standard pipeline should always look like this:
Load: image = cv2.imread('file.jpg') (Image is now BGR)
Process: Apply your OpenCV blurring, bitwise operations, or color masking directly to this BGR image.
Convert: image_rgb = cv2.cvtColor(processed_image, cv2.COLOR_BGR2RGB) (Image is now RGB)
Display: plt.imshow(image_rgb)

Phase 2: The ML Preprocessor (Gaussian Blur)
First, you will apply a standard blur often used as a preprocessing step for machine learning and deep learning models
. 3.  Apply the Filter: Use OpenCV's cv2.GaussianBlur() function. 4.  Analyze the Output: Observe how the Gaussian function averages nearby pixel values, giving closer pixels a higher weight
. This will result in a smooth, natural-looking blur

```python
imggaussian = cv2.GaussianBlur(img,(5,5),0)
```

```python
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original image in RGB format')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imggaussian, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Blurred image in RGB format')

plt.show()
```

Phase 3: The Static Cleaner (Median Blur)
Next, you will tackle a specific type of noise using a non-linear filter
. 5.  Apply the Filter: Use OpenCV's cv2.medianBlur() function. 6.  Analyze the Output: Instead of averaging, this filter replaces each pixel with the exact median value of its neighboring pixels
. If your test image has random black and white pixels (salt-and-pepper noise), you will see them disappear entirely while the actual edges of objects remain intact

```python
imgmedian = cv2.medianBlur(img,5)
```

```python
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('Original image in RGB format')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imgmedian, cv2.COLOR_BGR2RGB))
plt.title('Median Blurred image in RGB format')
```

Phase 4: The Detail Preserver (Bilateral Blur)
Now, you will use a more advanced technique for high-quality smoothing
. 7.  Apply the Filter: Use OpenCV's cv2.bilateralFilter() function. 8.  Analyze the Output: Because this filter calculates a weighted average based on both spatial distance and pixel intensity, it will blur areas that share similar colors while perfectly preserving sharp edges
. This gives you noise reduction without sacrificing important details 


cv2.bilateralFilter(src, d, sigmaColor, sigmaSpace)

```python
imgbilateral = cv2.bilateralFilter(img,5,75,75)
```

```python
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('Original image in RGB format')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(imgbilateral, cv2.COLOR_BGR2RGB))
plt.title('Bilateral Filtered image in RGB format')
```
